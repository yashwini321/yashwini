```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
cars_df=pd.read_csv('/content/archive (21).zip')
print(cars_df)
```

```python
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
print(a)
plt.xlabel('mpg')
plt.ylabel('weight')
plt.title("relationship between the weight of the car and the mpg")
plt.show()
```

```python
Trans=cars_df['am'].value_counts()
print(Trans)
Trans.plot.bar()
plt.xlabel('Transmission')
plt.title("Bar plot to check the frequency distribution of transmission")
plt.show()
```

```python
cars_df['mpg'].plot(kind='box')
plt.xlabel('mpg')
plt.title("Box and Whisker plot of mpg")
plt.show()
```

```python
min_value=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
max_value=np.max(cars_df['mpg'])
print('minimum_value:',min_value)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maximum_value:',max_value)
```
