```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ],

    "Sales": [
        120, 135, 150, 160,
        175, 190, 205, 220,
        235, 250, 270, 290
    ]
}
df = pd.DataFrame(data)
df["Month_Number"] = np.arange(1, 13)
print("\n================================================")
print("             HISTORICAL SALES DATA")
print("================================================")
print(df[["Month", "Sales"]])
X = df[["Month_Number"]]
y = df["Sales"]
model = LinearRegression()
model.fit(X, y)
df["Predicted_Sales"] = model.predict(X)
print("\n Actual prediction comparison")
    df[
        ["Month", "Sales", "Predicted_Sales"]
    ].to_string(index=False)
)
future_month_numbers = np.array(
    [13, 14, 15, 16, 17, 18]
).reshape(-1, 1)
future_predictions = model.predict(
    future_month_numbers
)
future_months = [
    "Jan-2027",
    "Feb-2027",
    "Mar-2027",
    "Apr-2027",
    "May-2027",
    "Jun-2027"
]
print("\n================================================")
print("           FUTURE SALES PREDICTION")
print("================================================")

for month, prediction in zip(
    future_months,
    future_predictions
):
    print(
        f"{month} : "
        f"{prediction:.2f} units"
    )
mae = mean_absolute_error(
    y,
    df["Predicted_Sales"]
)
mse = mean_squared_error(
    y,
    df["Predicted_Sales"]
)
rmse = np.sqrt(mse)
r2 = r2_score(
    y,
    df["Predicted_Sales"]
)
print("\n================================================")
print("             MODEL PERFORMANCE")
print("================================================")
print(f"Mean Absolute Error  : {mae:.2f}")
print(f"Mean Squared Error   : {mse:.2f}")
print(f"Root Mean Square Error : {rmse:.2f}")
print(f"R2 Score             : {r2:.4f}")
plt.figure(figsize=(10, 6))
plt.plot(
    df["Month"],
    df["Sales"],
    marker="o",
    linewidth=2,
    color="blue",
    label="Actual Sales"
)
plt.plot(
    df["Month"],
    df["Predicted_Sales"],
    marker="s",
    linestyle="--",
    linewidth=2,
    color="red",
    label="Predicted Sales"
)
plt.title(
    "Actual Sales vs Predicted Sales",
    fontsize=16
)
plt.xlabel(
    "Month",
    fontsize=12
)
plt.ylabel(
    "Sales (Units)",
    fontsize=12
)
plt.legend()
plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10, 6))
bars = plt.bar(
    future_months,
    future_predictions,
    color="orange",
    edgecolor="black"
)
plt.title(
    "Future Sales Prediction",
    fontsize=16
)
plt.xlabel(
    "Future Month",
    fontsize=12
)
plt.ylabel(
    "Predicted Sales (Units)",
    fontsize=12
)
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.6
)
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() +
        bar.get_width() / 2,
        height + 2,
        f"{height:.0f}",
        ha="center",
        fontsize=10
    )
plt.tight_layout()
plt.show()
print("\n================================================")
print("       SALES PREDICTION PROJECT COMPLETED")
print("================================================")
print("1. Historical sales data analyzed.")
print("2. Linear Regression model trained.")
print("3. Future sales predicted.")
print("4. Two graphs generated successfully.")
```
