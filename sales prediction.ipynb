```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
months = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
sales = np.array([120, 135, 150, 160, 175, 190, 205, 220])
model = LinearRegression()
model.fit(months, sales)
actual_predicted = model.predict(months)
future_months = np.array([9, 10, 11, 12]).reshape(-1, 1)
future_sales = model.predict(future_months)
print("Future Sales Predictions:")
for month, prediction in zip(future_months.flatten(), future_sales):
    print(f"Month {month}: {prediction:.2f} units")
plt.figure(figsize=(8, 5))
plt.plot(
    months,
    sales,
    marker="o",
    color="blue",
    label="Actual Sales"
)
plt.plot(
    months,
    actual_predicted,
    marker="o",
    color="green",
    label="Predicted Sales"
)
plt.title("Actual Sales vs Predicted Sales")
plt.xlabel("Month")
plt.ylabel("Sales (Units)")
plt.legend()
plt.grid(True)
plt.show()
plt.figure(figsize=(8, 5))
plt.bar(
    future_months.flatten(),
    future_sales,
    color="orange"
)
plt.title("Future Sales Prediction")
plt.xlabel("Future Month")
plt.ylabel("Predicted Sales (Units)")
plt.xticks(future_months.flatten())
plt.grid(axis="y")
plt.show()
```
